Q1. What is the average, median, and standard deviation of house prices?

```python
import pandas as pd

df = pd.read_csv('House Price India.csv')

mean_price = df['Price'].mean()
median_price = df['Price'].median()
std_price = df['Price'].std()

print(f"Mean Price: ${mean_price:,.2f}")
print(f"Median Price: ${median_price:,.2f}")
print(f"Standard Deviation: ${std_price:,.2f}")
```

Output / Results:

Average (Mean) Price: $\$538,932.22$
Median Price: $\$450,000.00$
Standard Deviation: $\$367,532.38$

Interpretation :
The average price is noticeably higher than the median price, which indicates a right-skewed distribution driven by a smaller group of high-priced luxury properties. The large standard deviation shows high variability and spread in property values across the dataset.

Q3. Check if the price data is skewed.

Q5.  What is the relationship between living area and price? (Use : Area without Basement)

```python
area_col = 'Area of the house(excluding basement)'
correlation = df[area_col].corr(df['Price'])
print(f"Correlation between Living Area (Excl. Basement) and Price: {correlation:.4f}")
```

Output / Results:

Correlation Coefficient: $0.6152$

Interpretation:
There is a moderate-to-strong positive correlation ($0.6152$) between the living area (excluding basement) and the house price. This demonstrates that as the functional living space of a home increases, its market price tends to increase reliably, making square footage one of the fundamental valuation metrics for real estate buyers.

Q6. Identify any anomalies where houses have high prices but low areas

```python
area_col = 'Area of the house(excluding basement)'

# Define threshold for high price (90th percentile) and low area (10th percentile)
high_price_threshold = df['Price'].quantile(0.90)
low_area_threshold = df[area_col].quantile(0.10)

# Filter anomaly records
anomalies = df[(df['Price'] > high_price_threshold) & (df[area_col] < low_area_threshold)]

print(f"High Price Threshold (Top 10%): ${high_price_threshold:,.2f}")
print(f"Low Area Threshold (Bottom 10%): {low_area_threshold:,.2f} sq ft")
print(f"Total Anomalies Found: {len(anomalies)}")

print(anomalies[['id', 'Price', area_col, 'grade of the house', 'waterfront present']])
```

Output / Results:

High Price Threshold (Top 10%): $\$882,991.00$
Low Area Threshold (Bottom 10%): $970.00\text{ sq ft}$
Total Anomalies Found: $1$
Anomaly Details:

ID: 6762811560
Price: $\$990,000.00$
Living Area (Excl. Basement): $930\text{ sq ft}$
Grade of House: $9$
Waterfront Present: $0$

Interpretation:
Anomalies where a property has a high price despite having a small living area occur primarily due to qualitative features rather than pure square footage. In this specific case, the property commands a premium price of $\$990,000$ for just $930\text{ sq ft}$ because it has an unusually high construction/design grade of $9$ out of $13$, indicating high-end luxury finishes or a prime location offset.

Q7. Compare average price based on number of floors and houses with or without waterfront.
